# Install System Dependencies

```python
import importlib.metadata as importlib_metadata
import os
import subprocess
import sys

TARGET_VERSIONS = {
    "numpy": "1.26.4",
    "pandas": "2.2.2",
}

restart_required = False

for package_name, target_version in TARGET_VERSIONS.items():
    try:
        current_version = importlib_metadata.version(package_name)
    except importlib_metadata.PackageNotFoundError:
        current_version = None

    if current_version != target_version:
        restart_required = True
        subprocess.check_call([
            sys.executable,
            "-m",
            "pip",
            "install",
            "--upgrade",
            "--force-reinstall",
            "--no-cache-dir",
            f"{package_name}=={target_version}",
        ])

if restart_required:
    print("Restarting runtime to finish loading updated numpy/pandas...")
    os.kill(os.getpid(), 9)
else:
    print("numpy and pandas already match the pinned versions.")
```
